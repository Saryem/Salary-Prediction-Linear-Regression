```python
def get_whiskers(col):
    q1 = col.quantile(0.25)
    q3 = col.quantile(0.75)
    IQR = q3 - q1
    lb = q1 - (1.5 * IQR)
    ub = q3 + (1.5 * IQR)
    return lb, ub

lb, ub = get_whiskers(df["selling_price"])
print("lower band", lb)
print("upper band", ub)

outliers = np.where((df["selling_price"] > ub) | (df["selling_price"] < lb))
outliers
```

```python
df = df.loc[
    (df["selling_price"] >= lb) &
    (df["selling_price"] <= ub)].copy()
print(df.shape)
```

Project Objective

The objective of this project was to develop a Machine Learning model capable of predicting employee salaries based on their years of professional experience using Linear Regression.

Model Evaluation and Findings

1. Mean Absolute Error (MAE): 729,372.70

The model produces an average absolute prediction error of approximately 729,373 salary units. This represents the average difference between actual and predicted salaries.

2. Mean Squared Error (MSE): 808,293,226,485.45

The MSE indicates the average squared prediction error. It penalises larger prediction errors more heavily and provides insight into the magnitude of model errors.

3. R² Score: 0.9547

The model achieved an R² score of approximately 95.47%, indicating that years of experience explains 95.47% of the observed variation in salaries within the evaluation dataset.

Overall Conclusion

The Linear Regression model demonstrates a strong linear relationship between years of professional experience and employee salary in the dataset.

The high R² score suggests that years of experience is a strong predictive feature for salary estimation in the observed data.

However, the model's predictive reliability should be further investigated through residual analysis, examination of regression assumptions, comparison of training and testing performance, and validation against unseen observations.

Further improvements could involve incorporating additional salary determinants such as educational qualifications, industry, job roles, professional certifications and geographical location.

The project demonstrates the application of supervised Machine Learning, data preprocessing, regression modelling and model evaluation in solving a practical salary prediction problem.
