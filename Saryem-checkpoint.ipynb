```python
def get_whiskers(col):
    q1 = col.quantile(0.25)
    q3 = col.quantile(0.75)
    IQR = q3 - q1
    lb = q1 - (1.5 * IQR)
    ub = q3 + (1.5 * IQR)
    return lb, ub

lb, ub = get_whiskers(df["selling_price"])
print("lower band", lb)
print("upper band", ub)

outliers = np.where((df["selling_price"] > ub) | (df["selling_price"] < lb))
outliers
```

```python
df = df.loc[
    (df["selling_price"] >= lb) &
    (df["selling_price"] <= ub)].copy()
print(df.shape)
```
